# RealSaS — DINOv2 Weight + Preprocess Authority V1

**Run All. Zero optimizer steps. No DINO feature extraction. No training.**

This notebook closes only the next preregistered authority gate:

1. require the promoted TRAIN512 native-1024 authority to be `512/512`;
2. clone exact `facebookresearch/dinov2` revision `7764ea0f...`;
3. derive the four official LVD-142M PTH URLs from that exact source;
4. download/cache the raw S/B/L/g PTH bytes to Drive and compute SHA-256 + size;
5. validate each PTH against its exact constructor with strict state-dict loading on the PyTorch `meta` device;
6. seal torch/torchvision/Pillow/NumPy versions;
7. seal a real native-1024 RGB preprocess fixture.

**Firewall:** this notebook never calls `forward` or `forward_features`, writes no candidate feature cache, constructs no optimizer, and performs zero scientific optimizer steps.


```python
# 0) Drive + fail-closed authorities
from google.colab import drive
drive.mount("/content/drive", force_remount=False)

from pathlib import Path
import hashlib, json, sys, subprocess

ROOT = Path("/content/drive/MyDrive/RealSaS_MASTER_CORPUS_1024_V3")
POST = ROOT / "reports" / "post_corpus_audit"
RUNNER = POST / "prepare_dino_weight_authority_v1.py"
OUT = ROOT / "reports" / "dino_controlled_preflight" / "DINO_WEIGHT_AUTHORITY_V1"
TRAIN_AUTH = POST / "DINO_TRAIN512_NATIVE1024_AUTHORITY_V2.json"

EXPECTED_RUNNER_SHA = "a65fb3378370c6c69fdbc19668ffc8346d96ee7e43f317215ebfa970bc51dbe4"
EXPECTED_TRAIN512_SET_SHA = "1958fa5ed80430ac8ae8f9e66f8d94bc5bfe89c74b5553d13b891c87fdefb2a2"

def sha256_file(p):
    h=hashlib.sha256()
    with open(p,"rb") as f:
        for b in iter(lambda:f.read(8<<20),b""): h.update(b)
    return h.hexdigest()

print("="*96, flush=True)
print("[DINO-WEIGHT-AUTH] NOTEBOOK START", flush=True)
print("[DINO-WEIGHT-AUTH] scientific_optimizer_steps=0", flush=True)
print("[DINO-WEIGHT-AUTH] candidate_feature_extraction=FORBIDDEN", flush=True)
print("="*96, flush=True)

if not ROOT.is_dir():
    raise RuntimeError(f"CORPUS_ROOT_MISSING: {ROOT}")
if not TRAIN_AUTH.is_file():
    raise RuntimeError(f"TRAIN512_AUTHORITY_MISSING: {TRAIN_AUTH}")
ta=json.loads(TRAIN_AUTH.read_text())
print("[TRAIN512] status:", ta.get("status"), flush=True)
print("[TRAIN512] ready:", ta.get("train512_ready_count"), "/", ta.get("train512_asset_count"), flush=True)
print("[TRAIN512] set_sha:", ta.get("train512_set_sha256"), flush=True)
if ta.get("status")!="PASS_512_OF_512" or ta.get("train512_ready_count")!=512 or ta.get("train512_asset_count")!=512:
    raise RuntimeError("TRAIN512_NOT_512_OF_512")
if ta.get("train512_set_sha256")!=EXPECTED_TRAIN512_SET_SHA:
    raise RuntimeError("TRAIN512_SET_SHA_DRIFT")

if not RUNNER.is_file():
    raise RuntimeError(f"RUNNER_MISSING: {RUNNER}")
got=sha256_file(RUNNER)
print("[runner] sha256:", got, flush=True)
if got!=EXPECTED_RUNNER_SHA:
    raise RuntimeError(f"RUNNER_SHA_DRIFT expected={EXPECTED_RUNNER_SHA} actual={got}")

OUT.mkdir(parents=True, exist_ok=True)
print("[output]", OUT, flush=True)
print("[DINO-WEIGHT-AUTH] AUTHORITY PREFLIGHT: PASS", flush=True)
```

```python
# 1) Execute sealed zero-step authority runner; stdout streams live
cmd=[
    sys.executable, str(RUNNER),
    "--corpus-root", str(ROOT),
    "--output-dir", str(OUT),
]
print("+", " ".join(cmd), flush=True)
p=subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
for line in p.stdout:
    print(line, end="", flush=True)
rc=p.wait()
print("[DINO-WEIGHT-AUTH] runner_exit_code:", rc, flush=True)
if rc!=0:
    raise RuntimeError(f"DINO_WEIGHT_AUTHORITY_RUNNER_FAILED exit={rc}")
```

```python
# 2) Final fail-closed readback
RESULT = OUT / "DINO_WEIGHT_AUTHORITY_V1.json"
if not RESULT.is_file():
    raise RuntimeError("DINO_WEIGHT_AUTHORITY_RESULT_MISSING")

r=json.loads(RESULT.read_text())
summary={
    "status": r.get("status"),
    "candidate_features_extracted": r.get("candidate_features_extracted"),
    "dino_training_started": r.get("dino_training_started"),
    "scientific_optimizer_steps": r.get("scientific_optimizer_steps"),
    "content_sha256": r.get("content_sha256"),
    "next_gate": r.get("next_gate"),
    "weights": {
        k: {
            "raw_pth_sha256": r["candidates"][k]["raw_pth_sha256"],
            "size_bytes": r["candidates"][k]["size_bytes"],
            "strict_load_pass": r["candidates"][k]["compatibility"]["strict_load_pass"],
        } for k in ["S","B","L","g"]
    },
    "preprocess_tensor_sha256": r["preprocess_fixture"]["preprocessed_tensor_sha256"],
}
print(json.dumps(summary, indent=2, sort_keys=True), flush=True)

if r.get("status")!="PASS_WEIGHT_BYTES_SOURCE_COMPAT_PREPROCESS_SEALED__FEATURE_OUTPUTS_UNOPENED":
    raise RuntimeError("WEIGHT_AUTHORITY_NOT_PASS")
if r.get("candidate_features_extracted") is not False:
    raise RuntimeError("FEATURE_FIREWALL_BREACH")
if r.get("dino_training_started") is not False or r.get("scientific_optimizer_steps")!=0:
    raise RuntimeError("TRAINING_FIREWALL_BREACH")
for k in ["S","B","L","g"]:
    if r["candidates"][k]["compatibility"]["strict_load_pass"] is not True:
        raise RuntimeError(f"{k}_STRICT_LOAD_NOT_PASS")

print("="*96, flush=True)
print("[DINO-WEIGHT-AUTH] COMPLETE: PASS", flush=True)
print("[DINO-WEIGHT-AUTH] FEATURE OUTPUTS REMAIN UNOPENED", flush=True)
print("[DINO-WEIGHT-AUTH] TRAINING REMAINS NOT AUTHORIZED", flush=True)
print("="*96, flush=True)
```
